> **Common Pitfall:** Methods like `.dropna()` and `.fillna()` return a *new* DataFrame and leave the original untouched unless you assign the result back (e.g., `populations = populations.dropna()`). If your "cleaning" seems to have no effect, check whether you forgot to capture the return value.

> **Common Pitfall:** When you call `.agg()` or `.groupby(...).agg()`, make sure you are grouping on the key you actually mean (e.g., `continent` vs. `region`). Grouping on the wrong column gives you a perfectly valid-looking table that answers the wrong question.

> **Common Pitfall:** `pd.pivot_table()` silently averages with `aggfunc='mean'` by default. If your summary numbers look surprising, confirm which aggregation function is being applied -- a "mean of counts" is rarely what you want.
